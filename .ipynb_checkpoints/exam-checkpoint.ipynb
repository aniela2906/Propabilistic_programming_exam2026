# Introduction

This report analyses a dataset of C programs generated by Large Language Models (LLMs). Each row corresponds to one generated program and contains information about the generating model, program size, cyclomatic complexity, and whether the program was classified as vulnerable.

The main research question is:

> **Are some LLMs less likely to generate vulnerable C code, and are code-size/complexity measures associated with vulnerability?**

The analysis investigates three hypotheses:

- **H1:** `llama2` is less likely to produce programs classified as `VULNERABLE` than the other LLMs.
- **H2:** Cyclomatic complexity is associated with the number of lines of code.
- **H3:** The number of lines of code increases the probability that a generated program is classified as `VULNERABLE`.

All hypotheses are analysed using Bayesian regression models. For binary vulnerability outcomes, I use logistic regression. For cyclomatic complexity, I use a Gaussian linear regression model after standardising the predictor variables.

Rows classified as `PARSE_ERROR` or `TIMEOUT` are excluded. These rows do not represent successfully generated programs for which vulnerability can be meaningfully modelled. The target variable is therefore built only from valid program classifications: `VULNERABLE` and `NON-VULNERABLE`.

#### Filtering out the `PARSE_ERROR` and `TIMEOUT`

#### 5. Standardise LOC and CC

```python
plt.figure(figsize=(8, 5))
plt.hist(df["num_lines"], bins=30)

plt.xlabel("Lines of code")
plt.ylabel("Frequency")
plt.title("Distribution of LOC before standardisation")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["cyclomatic_complexity"], bins=30)

plt.xlabel("Cyclomatic complexity")
plt.ylabel("Frequency")
plt.title("Distribution of cyclomatic complexity before standardisation")
plt.tight_layout()
plt.show()
```

```python
df["num_lines_std"] = (
    df["num_lines"] - df["num_lines"].mean()
) / df["num_lines"].std()

df["cyclomatic_complexity_std"] = (
    df["cyclomatic_complexity"] - df["cyclomatic_complexity"].mean()
) / df["cyclomatic_complexity"].std()

df[["num_lines", "num_lines_std", "cyclomatic_complexity", "cyclomatic_complexity_std"]].head()
```

The numerical predictors are standardised before modelling. This makes the regression coefficients easier to interpret and allows weakly informative priors to be used on a common scale.

```python
plt.figure(figsize=(8, 5))
plt.hist(df["num_lines_std"], bins=30)

plt.xlabel("Standardised lines of code")
plt.ylabel("Frequency")
plt.title("Distribution of LOC after standardisation")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["cyclomatic_complexity_std"], bins=30)

plt.xlabel("Standardised cyclomatic complexity")
plt.ylabel("Frequency")
plt.title("Distribution of cyclomatic complexity after standardisation")
plt.tight_layout()
plt.show()
```

Standardisation changes the scale of the variables but not the shape of their distributions. After standardisation, both variables have approximately mean 0 and standard deviation 1. This makes regression coefficients easier to compare and helps PyMC sample more efficiently.

## Bayesian modelling notation used in this report

For binary outcomes, the response variable is:

$$
y_i = \begin{cases}
1, & \text{if program } i \text{ is classified as VULNERABLE} \\
0, & \text{otherwise}
\end{cases}
$$

The logistic regression models use:

$$
y_i \sim \text{Bernoulli}(p_i)
$$

$$
\text{logit}(p_i) = \log \left(\frac{p_i}{1-p_i}\right) = \eta_i
$$

Equivalently:

$$
p_i = \frac{1}{1 + e^{-\eta_i}}
$$

This is useful because probabilities must stay between 0 and 1, while the linear predictor $\eta_i$ can take any real value. In plain language: the model does not predict vulnerability probability directly with a straight line, because straight lines can go below 0 or above 1. Logistic regression keeps the prediction inside the probability range, like putting bumpers on a bowling lane.
